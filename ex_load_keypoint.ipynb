```python
import os
import numpy as np
```

```python

npz_path = 'data01/NIA_SL_WORD0001_REAL01_D.npz'

data = np.load(npz_path)

keypoints = data['keypoints']    # (프레임 수, 키포인트 벡터 길이) 배열
label = data['label'].item()      # 스칼라 정수 (npz에 저장할 때 스칼라였으면 이렇게 꺼내야 함)

print("keypoints shape:", keypoints.shape)
print("label index:", label)
```

```python
folder_path = 'data03-norm'
npz_files = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.endswith('.npz')]
"""
[
    'data/NIA_SL_WORD0001_REAL01_D.npz',
    'data/NIA_SL_WORD1119_REAL01_R.npz',
    # ... 더 추가
]
"""
frame_size = []
for npz_path in npz_files:
    # npz_path = 'data01/NIA_SL_WORD0001_REAL01_D.npz'
    # npz_path = 'data01/NIA_SL_WORD1119_REAL01_R.npz'
    # ...

    data = np.load(npz_path)

    keypoints = data['keypoints']    # (프레임 수, 키포인트 벡터 길이) 배열
    label = data['label'].item()      # 스칼라 정수 (npz에 저장할 때 스칼라였으면 이렇게 꺼내야 함)
    frame_size.append(keypoints.shape[0])  # 프레임 수
    print("keypoints shape:", keypoints.shape)
    print("label index:", label)
```

```python
temp = sorted(frame_size)  # 최대 프레임 수
for i in temp:
    print(i)
```

```python
import pandas as pd
import os
import shutil
```

```python

# 1. 엑셀 파일 경로
excel_path = 'light_data_label/sign_language_labels_11.xlsx'  # 실제 경로로 변경

# 2. 원본 파일들이 있는 폴더 경로
source_folder = '/Users/hyeonji/Downloads/수어 영상/1.Training/11'

# 3. 복사할 대상 폴더 경로

target_folder = '/Users/hyeonji/Downloads/수어 영상/1.Training/11(F)'

# 대상 폴더가 없으면 생성
os.makedirs(target_folder, exist_ok=True)

# 4. 엑셀 파일 읽기
df = pd.read_excel(excel_path)

# 5. 엑셀에서 파일 이름 리스트 가져오기
file_names = df['file_name'].tolist()  # 컬럼명이 file_name이라고 가정, 실제 컬럼명에 맞게 수정

# 6. 파일 복사
for file_name in file_names:
    source_path = os.path.join(source_folder, file_name)
    target_path = os.path.join(target_folder, file_name)

    if os.path.exists(source_path):
        shutil.copy2(source_path, target_path)
        print(f'Copied {file_name}')
    else:
        print(f'File not found: {file_name}')

print('File copying completed.')
```

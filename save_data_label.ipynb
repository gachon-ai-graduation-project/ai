```python
"""
    idx2word2에 적힌 클래스를 갖는 파일들만 엑셀에 기록하기
    파일명, 라벨, 라벨인덱스
    정보가 저장됨

    data_label 폴더에 엑셀로 저장됨
"""
```

```python
import json
import os
import pandas as pd
```

```python
def json_to_excel(json_paths, output_excel_path):
    # 2. 불러오기
    with open('word2idx2.json', 'r', encoding='utf-8') as f:
        word2idx = json.load(f)

    with open('idx2word2.json', 'r', encoding='utf-8') as f:
        idx2word = json.load(f)
        
    # 2) 엑셀에 들어갈 데이터 준비
    rows = []
    for path in json_paths:
        print(f"Processing {path}...")
        with open(path, 'r', encoding='utf-8') as f:
            data = json.load(f)
            file_name = data.get("metaData", {}).get("name", "")
            
            if file_name.endswith("_F.mp4"): # 정면 데이터만 사용
                # data 안에 여러 개의 item 있을 수 있음
                for item in data.get("data", []):
                    for attr in item.get("attributes", []):
                        label = attr.get("name", "")
                        idx = word2idx.get(label, word2idx["<UNK>"])
                        if idx == 1:
                            continue  # Skip if label is "<UNK>"
                        rows.append({
                            "file_name": file_name,
                            "label": label,
                            "label_index": idx
                        })
    
    # 3) DataFrame 생성 및 엑셀 저장
    print(len(rows))
    df = pd.DataFrame(rows, columns=["file_name", "label", "label_index"])
    df.to_excel(output_excel_path, index=False)
    print(f"엑셀 파일 저장 완료: {output_excel_path}")

# 사용 예시
if __name__ == "__main__":
    # JSON 파일 경로들 리스트로 준비
    json_folder = "/Users/hyeonji/Downloads/수어 영상/1.Training/morpheme/11"
    json_files = [os.path.join(json_folder, f) for f in os.listdir(json_folder) if f.endswith(".json")]
    print(f"JSON 파일 개수: {len(json_files)}")

    output_path = "light_data_label/sign_language_labels_11.xlsx"
    json_to_excel(json_files, output_path)
```

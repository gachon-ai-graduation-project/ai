```python
"""
morpheme/XX 폴더엥 있는 파일을 읽어서 단어 이름을 가져온다
단어 이름을 모아 word2idx, idx2word를 만든다.
"""
```

```python
import json
import os
import pandas as pd

def build_vocab(label_json_paths):
    vocab = set()
    for label_path in label_json_paths:
        with open(label_path, 'r', encoding='utf-8') as f:
            data = json.load(f)
            for item in data.get('data', []):
                for attr in item.get('attributes', []):
                    vocab.add(attr['name'])
    return sorted(list(vocab))

def create_word_mappings(vocab):
    word2idx = {"<PAD>":0, "<UNK>":1}
    idx2word = {0:"<PAD>", 1:"<UNK>"}
    for i, word in enumerate(vocab, start=2):
        word2idx[word] = i
        idx2word[i] = word
    return word2idx, idx2word

def create_json(json_paths):
    # 1) 단어 집합 구축
    vocab = build_vocab(json_paths)
    word2idx, idx2word = create_word_mappings(vocab)
        
    # JSON은 키가 문자열만 가능하니까 idx2word의 키는 str로 변환
    with open('word2idx2.json', 'w', encoding='utf-8') as f:
        json.dump(word2idx, f, ensure_ascii=False, indent=4)

    with open('idx2word2.json', 'w', encoding='utf-8') as f:
        json.dump({str(k): v for k,v in idx2word.items()}, f, ensure_ascii=False, indent=4)
    

# 사용 예시
if __name__ == "__main__":
    # JSON 파일 경로들 리스트로 준비
    json_folder = "/Users/hyeonji/Downloads/수어 영상/1.Training/morpheme/03/"
    json_files = [os.path.join(json_folder, f) for f in os.listdir(json_folder) if f.endswith(".json")]
    
    create_json(json_files)
```
